# <img src="tutorial-icons/theory.png" alt="Theory" width="44" style="vertical-align:middle; margin-right:10px;"> Theory Foundation: Problems, Search, and Verification

Before using Claude, separate the **problem** from the **first solution idea**. A problem can be represented as a current state $s_0$, a desired goal set $G$, a set of allowed actions $A$, and constraints $C$. A solution is a sequence of actions

$$
\pi=(a_1,a_2,\ldots,a_T)
$$

that moves the work from $s_0$ to a state $s_T\in G$ without violating $C$. This is why Pólya's stages are more than a checklist: understanding defines $s_0$, $G$, and $C$; planning searches for a plausible $\pi$; carrying out the plan executes it; and looking back tests whether the result truly belongs to $G$.

When several solutions are possible, we often evaluate an objective or loss. A simple form is

$$
J(\pi)=\text{error}(\pi)+\lambda\,\text{cost}(\pi)+\gamma\,\text{risk}(\pi),
$$

where $\lambda$ and $\gamma$ make priorities explicit. There is no universal choice of weights: deciding what error, effort, and harm matter is a human and domain judgment. Claude can generate candidate plans quickly, but fluent output is not evidence that the constraints were understood or that the objective was appropriate.

Verification should therefore use at least three layers: **internal checks** (does the reasoning or code agree with itself?), **external checks** (does it agree with known facts, tests, or data?), and **consequence checks** (what happens if it is wrong?). In the Galton-board example, the mathematical reference is a binomial distribution,

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k},\qquad E[X]=np,\qquad \operatorname{Var}(X)=np(1-p).
$$

Simulation results should approach—not exactly reproduce—these quantities as the number of trials grows. The central habit for the whole series is: **state the problem, expose assumptions, predict before running, test the result, and record what remains uncertain**.

### Visual intuition: what is a Galton board?

<img src="tutorial-icons/galton_board_intro.gif" alt="Animated Galton board showing balls making repeated left/right decisions and accumulating in bins" width="680">

If this is your first Galton board, follow **one ball**: at each peg it makes one left/right decision. After all rows, the final bin is determined by the **number of right moves**. The animation is conceptual; the coded simulation below uses the tutorial’s stated settings.

[Optional external visual: Mathspig Blog — Galton Board](https://mathspig.wordpress.com/tag/galton-board/)


A Galton board contains rows of pegs. At each row, a ball moves one position left or right. If both directions are equally likely, the final bin is determined by the number of right moves.

For this simulation, the exact theoretical distribution is **binomial**:

$$
K \sim \operatorname{Binomial}(n_{\text{rows}},p_{\text{right}})
$$

With enough balls, the binomial counts form a roughly symmetric, bell-shaped pattern when `p_right = 0.50`. That shape can resemble a Gaussian, or normal, curve. The important distinction is that the simulated final bins are discrete and exactly modeled by a binomial distribution; the Gaussian curve is only an approximation.

The goal is not merely to obtain a plot. The goal is to carry out a plan in small, checkable steps.

### <img src="tutorial-icons/reference_solution.png" alt="Reference Solution" width="30" style="vertical-align:middle; margin-right:8px;"> Reference Solution for Task 3

Compare this solution with Claude’s code only after making your own attempt. The exact binomial expectation is the main theoretical comparison. The Gaussian line is included to show why the discrete binomial pattern can look bell-shaped.


````python
import math
import numpy as np
import matplotlib.pyplot as plt

n_rows = 12
n_balls = 1000
p_right = 0.50
random_seed = 1099

rng = np.random.default_rng(random_seed)

# One row per ball and one column per peg row.
moves = rng.binomial(
    n=1,
    p=p_right,
    size=(n_balls, n_rows),
)

# The final bin is the number of right moves.
final_bins = moves.sum(axis=1)
bin_numbers = np.arange(n_rows + 1)
observed_counts = np.bincount(final_bins, minlength=n_rows + 1)

# Exact binomial expectation.
binomial_probabilities = np.array([
    math.comb(n_rows, k)
    * (p_right ** k)
    * ((1 - p_right) ** (n_rows - k))
    for k in bin_numbers
])
expected_binomial_counts = n_balls * binomial_probabilities

# Gaussian approximation evaluated at the integer bin locations.
theoretical_mean = n_rows * p_right
theoretical_variance = n_rows * p_right * (1 - p_right)
theoretical_sd = np.sqrt(theoretical_variance)

gaussian_approximation = (
    n_balls
    * (1 / np.sqrt(2 * np.pi * theoretical_variance))
    * np.exp(
        -((bin_numbers - theoretical_mean) ** 2)
        / (2 * theoretical_variance)
    )
)

assert final_bins.min() >= 0
assert final_bins.max() <= n_rows
assert observed_counts.sum() == n_balls
assert len(observed_counts) == n_rows + 1

observed_mean = final_bins.mean()
observed_variance = final_bins.var()

print(f"Total simulated balls: {observed_counts.sum()}")
print(f"Observed mean final bin: {observed_mean:.3f}")
print(f"Observed variance: {observed_variance:.3f}")
print(f"Theoretical mean: {theoretical_mean:.3f}")
print(f"Theoretical variance: {theoretical_variance:.3f}")

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(
    bin_numbers,
    observed_counts,
    alpha=0.65,
    label="Observed simulation",
)
ax.plot(
    bin_numbers,
    expected_binomial_counts,
    marker="o",
    linewidth=2,
    label="Exact binomial expectation",
)
ax.plot(
    bin_numbers,
    gaussian_approximation,
    linestyle="--",
    linewidth=2,
    label="Gaussian approximation",
)
ax.set_title("Galton Board: Observed Counts and Theoretical Shapes")
ax.set_xlabel("Final bin = number of right moves")
ax.set_ylabel("Number of balls")
ax.set_xticks(bin_numbers)
ax.legend()
plt.show()
````

## Lesson Learned From the Galton Board

Looking back reveals more than whether the plot “looks right.”

1. A complex-looking aggregate pattern can emerge from many simple random decisions.
2. The exact distribution is binomial because the final bin counts right moves across a fixed number of independent trials.
3. A Gaussian curve can approximate the overall bell shape, but it is not the exact discrete model.
4. Verification is stronger when simulation results are compared with known theoretical quantities.
5. Increasing the number of balls reduces visible simulation noise; changing the number of rows changes the distribution itself.
6. The reusable lesson is to identify the underlying structure, derive expected behavior, and test generated code against that expectation.

A solution can run successfully and still fail to represent the intended problem. Looking back is where that difference becomes visible.


## <img src="tutorial-icons/theory.png" alt="Theory" width="36" style="vertical-align:middle; margin-right:9px;"> Theory: Looking Back Creates Generalization

Verification asks whether this solution is correct for the chosen inputs. Generalization asks which relationships remain true when the inputs change. Keeping those questions separate prevents one successful run from becoming an unsupported universal claim.

### Deeper explanation

Generalization extracts a reusable rule while preserving its assumptions. A useful record names inputs, outputs, constraints, failure cases, and evidence. If a solution works on examples $D$, the conclusion is not “it always works” but “it passed these tests under these conditions.” Looking back also separates accidental success from causal mechanism: change one assumption, rerun a test, and observe whether the explanation predicts the change. This creates knowledge that can guide the next problem rather than merely a saved answer.

# AI for Social Good: Start With the Human Problem, Not the Tool

The earlier IE 1171 course repeatedly showed that a technically strong model can still fail if the **problem was scoped badly**, the wrong people were consulted, or the result cannot survive deployment in the real world. Social-good projects have included public health, food security, conservation, civic services, misinformation, and poverty alleviation—but the useful pattern is the same: begin with the need and the people affected, not with a favorite algorithm.

Use Pólya’s cycle as a social-good discipline:

- **Understand the problem:** Who experiences the problem? Who defined success? What constraints, rights, and local knowledge matter?
- **Devise a plan:** Compare several interventions, including non-AI options. Decide what evidence would make the plan acceptable.
- **Carry out the plan:** Keep assumptions, data choices, AI contributions, and uncertainty visible.
- **Look back:** Check what happened to people after deployment, including unintended consequences and groups that may have been missed.

Before accepting an AI-assisted answer, ask:

1. Is this the real social problem, or only the easiest thing to measure?
2. Which stakeholders or domain experts would disagree with our framing?
3. What could go wrong outside the notebook even if the code is correct?
4. Who can challenge the result or ask for a different decision?
5. What human remains accountable?

> **Social-good principle:** A good algorithm cannot rescue a badly framed problem. Human-centered scoping, stakeholder knowledge, and post-deployment learning are part of the technical work.

# Part 6: Productive Disagreement with Claude

Claude may present one plausible path as though it were the only path.

Ask for disagreement deliberately:

```text
Challenge the current plan.

Provide:
1. the strongest argument against it;
2. an alternative method;
3. a case where the alternative would be better;
4. evidence that would distinguish the two plans;
5. one assumption shared by both plans that may still be wrong.
```

The goal is not endless debate. The goal is to expose hidden choices before they become embedded in code.


# Part 7: Generalization — Turn One Solution Into a Reusable Method

Pólya’s “look back” stage asks whether the completed solution can teach us something more general.

For the Galton board, use this direct question:

> **Which properties of the solution remain true for any values of `n_rows`, `n_balls`, and `p_right`, and which properties depend on the chosen values?**

Ask Claude to act as a problem-solving partner:

```text
Use the completed Galton board solution as evidence. Help me generalize it,
not merely summarize it.

Answer these questions directly:

1. What is the general relationship between a ball's final bin and its number
   of right moves?
2. What exact probability distribution describes the final bin for arbitrary
   n_rows and p_right?
3. Which outputs change when n_balls changes, and which theoretical quantities
   do not?
4. Which outputs change when n_rows changes?
5. What changes when p_right is not 0.50?
6. Which assertions or invariants should remain valid in every version?
7. Under what conditions should the binomial distribution look approximately
   Gaussian?
8. Give one different real-world problem with the same repeated binary-trial
   structure, and map each Galton board object to that problem.
9. Ask me one question that tests whether I understand the generalization.

Do not write new code until the generalized structure is clear.
```

> **Domain expertise required:** The questions below are examples, not a complete checklist. Add a check based on the real application; if you lack that expertise, involve someone who has it.

- **Your own domain question:** What could be wrong here that a generic AI system or checklist would be unlikely to notice?


A useful generalization should:

- state the structure precisely;
- separate exact facts from approximations;
- identify invariants;
- explain which assumptions are necessary;
- transfer the method without pretending every new problem is identical.

# Part 8: Pólya’s Heuristic Toolbox

When a direct plan is not obvious, try a question that changes how the problem is represented.

| Heuristic | Direct question | Galton board example |
|---|---|---|
| **Analogy** | Have I solved a problem with the same structure? | Connect the final bin to counting successes in repeated trials. |
| **Specialization** | Can I solve a smaller case completely? | Use 2 rows and list every possible path. |
| **Auxiliary problem** | What smaller result would unlock the next step? | Verify that one ball’s final bin equals its number of right moves. |
| **Work backward** | What must be true immediately before the goal? | To finish in bin 8 after 12 rows, the ball must have made 8 right moves. |
| **Decompose and recombine** | Which parts can be checked separately? | Separate simulation, counting, theory, plotting, and validation. |
| **Generalize** | What remains true after the numerical settings change? | Keep the binomial structure while varying rows, balls, or probability. |
| **Use all conditions** | Did the solution use every stated constraint? | Confirm the code uses the requested rows, balls, probability, and seed. |

Ask Claude to apply **two different heuristics** before accepting a new plan. Different heuristics can reveal different assumptions and failure points.


# Part 9: Decomposing Difficult Problems

When a task feels too large, use three passes:

## Pass 1 — Map the problem

Ask Claude to identify components, dependencies, unknowns, and decision points.

## Pass 2 — Isolate the next verifiable step

Choose one step that produces a checkable output.

## Pass 3 — Integrate

After the parts work individually, inspect how assumptions and errors travel between them.

This is why Tutorials 1–3 separate data understanding, regression, and model selection rather than asking Claude to create the entire project in one response.


# Part 10: Common Failure Modes

| Failure mode | Why it happens | Better response |
|---|---|---|
| Method-first prompting | The user names an algorithm before defining the problem | Return to Pólya Stage 1 |
| One giant prompt | Too many hidden decisions are generated at once | Divide work into verifiable tasks |
| Automation bias | Polished output feels trustworthy | Require tests and independent evidence |
| Confirmation seeking | Claude is asked to approve an existing answer | Ask for counterarguments and failure tests |
| Context loss | Variable meanings are omitted | Provide codebook definitions and units |
| Silent data leakage | Evaluation data influence modeling choices | Freeze a test set before model development |
| Causal overreach | Association language becomes intervention advice | Match claims to study design |
| Social-context loss | Accuracy becomes the only criterion | Examine stakeholders, subgroup errors, and intended use |
