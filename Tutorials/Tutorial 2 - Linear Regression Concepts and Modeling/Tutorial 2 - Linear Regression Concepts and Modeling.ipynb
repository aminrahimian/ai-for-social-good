# <img src="tutorial-icons/theory.png" alt="Theory" width="44" style="vertical-align:middle; margin-right:10px;"> Theory Foundation: Linear Models, Estimation, and Generalization

Multiple linear regression assumes that the conditional mean of a quantitative response is a linear combination of predictors:

$$
Y_i=\beta_0+\beta_1X_{i1}+\cdots+\beta_pX_{ip}+\varepsilon_i,
\qquad E(\varepsilon_i\mid X_i)=0.
$$

The fitted coefficients minimize the residual sum of squares

$$
RSS(\beta)=\sum_{i=1}^{n}(y_i-\hat y_i)^2
=\lVert y-X\beta\rVert_2^2.
$$

When $X^TX$ is invertible, the ordinary least-squares solution is $\hat\beta=(X^TX)^{-1}X^Ty$. This formula explains why exact or severe multicollinearity is difficult: nearly dependent columns make $X^TX$ unstable, so small changes in the data can produce large coefficient changes.

A residual is $e_i=y_i-\hat y_i$. Residual plots investigate whether the mean relationship is approximately linear, whether error variance is roughly constant, and whether a few observations dominate the fit. Normal errors are needed for familiar small-sample $t$ and $F$ inference, but not for the mechanical act of minimizing RSS. Independence, appropriate sampling, and the absence of leakage are separate requirements.

For prediction, training fit is not enough. Common test metrics are

$$
MSE=\frac{1}{n_{test}}\sum_i(y_i-\hat y_i)^2,
\qquad RMSE=\sqrt{MSE},
\qquad MAE=\frac{1}{n_{test}}\sum_i|y_i-\hat y_i|.
$$

$R^2=1-RSS/TSS$ measures improvement over predicting the sample mean on the evaluated data; it does not prove causality, fairness, or useful error size. Coefficients are conditional associations: $\beta_j$ is the expected change in $Y$ for a one-unit increase in $X_j$ **holding the other included predictors fixed**. That interpretation becomes fragile when the change is impossible, unsupported by the data, or confused with intervention.

# Part 1: Simple Linear Regression

Regression estimates the average relationship between one or more predictors and a response.

## Response variable

The response is the outcome we want to explain or predict.

In this tutorial:

$$
Y_i = PROPr_i
$$

## Predictor variable

A predictor is a variable used to estimate the response.

For the first model:

$$
X_i = StressBurden_i
$$

## Simple linear-regression equation

$$
\hat{Y}_i = \hat{\beta}_0 + \hat{\beta}_1 X_i
$$

Where:

- $\hat{Y}_i$ is the predicted response for observation $i$;
- $\hat{\beta}_0$ is the fitted intercept;
- $\hat{\beta}_1$ is the fitted slope;
- $X_i$ is the predictor value.

````python
from sklearn.linear_model import LinearRegression

simple_df = analysis_ready[
    ["PROPr", "StressBurden"]
].dropna().copy()

X_simple = simple_df[["StressBurden"]]
y_simple = simple_df["PROPr"]

simple_model = LinearRegression()
simple_model.fit(X_simple, y_simple)

simple_intercept = simple_model.intercept_
simple_slope = simple_model.coef_[0]
simple_r2 = simple_model.score(X_simple, y_simple)

print(f"Intercept: {simple_intercept:.4f}")
print(f"StressBurden slope: {simple_slope:.4f}")
print(f"R^2: {simple_r2:.4f}")

plt.figure(figsize=(8, 5))

plt.scatter(
    simple_df["StressBurden"],
    simple_df["PROPr"],
    alpha=0.25,
)

x_line = np.linspace(
    simple_df["StressBurden"].min(),
    simple_df["StressBurden"].max(),
    100,
).reshape(-1, 1)

y_line = simple_model.predict(x_line)

plt.plot(x_line, y_line)

plt.title("Simple Linear Regression: Stress Burden and PROPr")
plt.xlabel("StressBurden: higher means more frequent stress")
plt.ylabel("PROPr: higher means better health-related quality of life")
plt.tight_layout()
plt.show()
````

# Part 2: Predictions and Residuals

A fitted model generates a prediction for each observation.

A residual is the difference between the observed and predicted values:

$$
e_i = Y_i - \hat{Y}_i
$$

A residual tells us how far the model missed.

| Residual | Meaning |
|---|---|
| Positive | Observed `PROPr` was higher than predicted |
| Negative | Observed `PROPr` was lower than predicted |
| Near zero | Prediction was close |

Residual plots help us identify patterns the model did not capture.


## <img src="tutorial-icons/theory.png" alt="Theory" width="36" style="vertical-align:middle; margin-right:9px;"> Theory: Residuals Connect the Equation to Model Error

A fitted value is the model’s estimate for one row; a residual is the observed response minus that estimate. Least squares chooses coefficients that minimize the sum of squared training residuals. A small training residual is not proof of good generalization, so residual patterns and held-out performance must also be examined.

### Deeper explanation

Residuals satisfy $y_i=\hat y_i+e_i$. A random-looking residual cloud supports—but cannot prove—the chosen mean form. Curvature suggests missing nonlinear structure; a funnel suggests nonconstant variance; clusters may signal omitted groups; extreme leverage points can alter the fitted line. Residuals are observed errors around fitted values, while the unobserved $\varepsilon_i$ are population errors. Confusing the two leads to overconfident diagnostics.

# Part 4: Train-Test Evaluation

A model can appear accurate when evaluated on the same data used to fit it.

To obtain a more honest evaluation, split the rows into:

- **training set:** used to fit the model;
- **test set:** held back until evaluation.

We will use:

- 80% training data;
- 20% test data;
- `random_state=1099` so the split is reproducible.


## <img src="tutorial-icons/theory.png" alt="Theory" width="36" style="vertical-align:middle; margin-right:9px;"> Theory: Test Error Estimates Generalization

ISLP distinguishes training error from test error. Training rows determine the coefficients; test rows approximate how the completed model performs on unseen observations. The test set must remain outside feature choices and tuning decisions, or it stops being an honest final check.

### Deeper explanation

If model choices use the test outcomes, the reported test error becomes optimistically biased. Conceptually, expected generalization error is $E_{(X,Y)\sim P}[L(Y,\hat f(X))]$, while a finite test mean estimates it with sampling uncertainty. The estimate applies only when future cases resemble the test-generating distribution. Distribution shift can invalidate an otherwise honest split.


## Evaluation Metrics

### Mean Absolute Error

$$
MAE =
\frac{1}{n}
\sum_{i=1}^{n}
|Y_i - \hat{Y}_i|
$$

MAE is the average absolute prediction error.

### Root Mean Squared Error

$$
RMSE =
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(Y_i - \hat{Y}_i)^2
}
$$

RMSE gives more weight to larger errors.

### R-squared

$$
R^2 =
1 -
\frac{
\sum_{i=1}^{n}(Y_i - \hat{Y}_i)^2
}{
\sum_{i=1}^{n}(Y_i - \bar{Y})^2
}
$$

$R^2$ describes how much variation the model explains relative to predicting the mean.

# AI for Social Good: Regression Can Inform Decisions Without Defining Fairness

Regression can support public-interest work such as studying health outcomes, housing conditions, service demand, or how resources relate to community needs. But a good fit does not tell us whether a downstream decision is fair or appropriate.

Useful ethical ideas from the earlier course help separate **prediction quality** from **decision quality**:

- historical associations can reflect existing inequality;
- apparently neutral predictors can act as proxies for structural conditions;
- overall RMSE or $R^2$ can hide groups with larger or systematically directed errors;
- removing a protected attribute does not automatically remove its information from the model;
- fairness also depends on what happens **before and beyond the algorithm**—problem definition, data collection, policy rules, human discretion, and appeal.

For this notebook, the model is an educational analysis of associations. If a similar model were used for resource allocation or access to services, a domain expert would need to determine which errors are consequential and whether the outcome being predicted is even the right target.

Before deployment, ask:

1. Which groups have the largest errors and the smallest sample sizes?
2. Are residuals systematically high or low for any group?
3. Does the model describe an association, or is someone treating it as a causal effect?
4. Could a policy based on the prediction reinforce the pattern used to train the model?
5. What explanation, review, or appeal would affected people receive?

> **Social-good principle:** Regression can inform a decision; it cannot by itself define what a fair decision should be.

````python
social_audit = model_df.loc[
    y_test.index,
    ["INCOME", "EDUC4"],
].copy()

social_audit["actual_PROPr"] = y_test
social_audit["predicted_PROPr"] = test_predictions
social_audit["residual"] = (
    social_audit["actual_PROPr"] - social_audit["predicted_PROPr"]
)
social_audit["absolute_error"] = social_audit["residual"].abs()

income_error_table = (
    social_audit
    .groupby("INCOME", dropna=False)
    .agg(
        count=("absolute_error", "size"),
        mean_residual=("residual", "mean"),
        MAE=("absolute_error", "mean"),
    )
    .reset_index()
    .sort_values("INCOME")
)

education_error_table = (
    social_audit
    .groupby("EDUC4", dropna=False)
    .agg(
        count=("absolute_error", "size"),
        mean_residual=("residual", "mean"),
        MAE=("absolute_error", "mean"),
    )
    .reset_index()
    .sort_values("EDUC4")
)

print("Test error by income category:")
display(income_error_table)

print("\nTest error by education category:")
display(education_error_table)

print(
    "\nWarning: These are descriptive group summaries. "
    "Small groups, sampling variation, measurement choices, and omitted "
    "context must be investigated before making fairness claims."
)
````
